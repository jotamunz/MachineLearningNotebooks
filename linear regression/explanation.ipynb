The linear regression equation is defined as $ y = mx + b $, which is equivalent to $ y' = b + w_1x_1 $

- $ y $ is miles per gallon or the label (the value to predict)
- $ m $ is the slope of the line or a weight (a trainable parameter)
- $ x $ is pounds or a feature (input value)
- $ b $ is the y-intercept or the bias (a trainable parameter).

A more complex linear regression model may rely on more than one feature:

$ y' = b + \sum_{i=1}^{n} w_i x_i $


## Loss Functions

loss measures the difference between the predicted and actual values. Models are trained by minimizing the loss.

![Figure 8. Loss lines connect the data points to the
model.](https://developers.google.com/static/machine-learning/crash-course/linear-regression/images/loss-lines.png)

In linear regression, there are five main types of loss:

| Loss type | Definition | Equation |
|---|---|---|
| $L_1$ loss | The sum of the absolute values of the difference between the predicted values and the actual values. | $\sum \lvert \text{actual value} - \text{predicted value} \rvert$ |
| Mean absolute error (MAE) | The average of $L_1$ losses across a set of *N* examples. | $\frac{1}{N} \sum \lvert \text{actual value} - \text{predicted value} \rvert$ |
| $L_2$ loss | The sum of the squared difference between the predicted values and the actual values. | $\sum (\text{actual value} - \text{predicted value})^2$ |
| Mean squared error (MSE) | The average of $L_2$ losses across a set of *N* examples. | $\frac{1}{N} \sum (\text{actual value} - \text{predicted value})^2$ |
| Root mean squared error (RMSE) | The square root of the mean squared error (MSE). | $\sqrt{\frac{1}{N} \sum (\text{actual value} - \text{predicted value})^2}$ |

The functional difference between $L_1$ loss and $L_2$ loss (or between MAE/RMSE and MSE) is squaring. When the difference between the prediction and label is large, squaring makes the loss even larger. When the difference is small (less than 1), squaring makes the loss even smaller.

Loss metrics like MAE and RMSE may be preferable to $L_2$ loss or MSE in some use cases because they tend to be more human-interpretable, as they measure error using the same scale as the model's predicted value.

### Choosing a loss

MSE moves the model more toward the outliers, while MAE doesn't. $L_2$ loss incurs a much higher penalty for an outlier than $L_1$ loss. 

**Choose MSE:**

- If you want to heavily penalize large errors.
- If you believe the outliers are important and indicative of true data variance that the model should account for.

> **Note:** The mathematical properties of MSE often make optimization smoother. Root Mean Squared Error (RMSE) is often used to get the error back into the same units as the label.

**Choose MAE:**

- If your dataset has significant outliers that you don't want to overly influence the model. MAE is more robust.
- If you prefer a loss function that is more directly interpretable as the average error magnitude.

## Gradient descent

Iteratively finds the weights and bias that produce the model with the lowest loss. The model begins training with randomized weights and biases near zero, and then repeats the following steps:

1. Calculate the loss with the current weight and bias.
2. Determine the direction to move the weights and bias that reduce loss.
3. Move the weight and bias values a small amount (called learning rate) in that direction.
4. Return to step one and repeat the process until the model can't reduce the loss any further.

The loss functions for linear models always produce a convex surface. A linear model converges when it's found the minimum loss through the optimal weights and bias. In practice these values are approximations.

## Hyperparameters

Parameters are the values that are part of the model itself and it calculates during training, while hyperparameters are variables that control the aspects of training and are set by you.

### Learning rate

Determines the magnitude of the changes to make to the weights and bias during each step of gradient descent. If the learning rate is too low, the model can take a long time to converge. However, if the learning rate is too high, the model never converges, but instead bounces around the weights and bias that minimize the loss.

### Batch size

The number of examples the model processes before updating its weights and bias. 

- **Full batch**: A batch is composed of all the exampled, meaning it looks at all of the examples before updating the parameters.

- **Stochastic gradient descent (SGD)**: A batch has a size of 1, meaning it updates the parameters after every example. The stochastic part comes from taking the example at random. SGD is very noisy

- **Mini-batch stochastic gradient descent (mini-batch SGD)**: A batch has a size between 1 and the total number of examples, which it groups randomly and averages the gradients from the batch.

### Epochs

During training, the model has processed every example in the training set *once*. The required number of epochs depends on how fast the model converges.

| Batch type | When weights and bias updates occur |
|---|---|
| Full batch | After the model looks at all the examples in the dataset. For instance, if a dataset contains 1,000 examples and the model trains for 20 epochs, the model updates the weights and bias 20 times, once per epoch. |
| Stochastic gradient descent | After the model looks at a single example from the dataset. For instance, if a dataset contains 1,000 examples and trains for 20 epochs, the model updates the weights and bias 20,000 times. |
| Mini-batch stochastic gradient descent | After the model looks at the examples in each batch. For instance, if a dataset contains 1,000 examples, and the batch size is 100, and the model trains for 20 epochs, the model updates the weights and bias 200 times. |
